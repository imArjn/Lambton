# Dataset Features Table - Canadian Cultural Events

| Feature | Data Type | Description | Purpose |
|---------|-----------|-------------|---------|
| post_id | String | Unique identifier for each social media post | Primary key for individual post records |
| event_name | String | Name of the Canadian Heritage cultural event | Group posts by specific events for comparison |
| event_type | String | Category of cultural event (National Holiday, Cultural Awareness, etc.) | Analyze performance by event category |
| event_date | DateTime | Official date of the cultural event | Time-based analysis and campaign timing |
| campaign_budget_cad | Integer | Total budget allocated for the event campaign in CAD | Cost analysis and ROI calculations |
| campaign_duration_days | Integer | Number of days the campaign ran | Calculate campaign intensity and duration impact |
| location_type | String | Geographic scope of event (National, Regional, Provincial) | Analyze performance by geographic reach |
| platform | String | Social media platform name (Facebook, Instagram, Twitter, YouTube, TikTok) | Compare performance across different platforms |
| platform_demographic | String | Primary age demographic for each platform | Match event content with target audiences |
| content_type | String | Type of content posted (image, video, carousel, story, text) | Analyze which content formats drive engagement |
| post_date | DateTime | Date when the social media post was published | Track engagement patterns over time |
| impressions | Integer | Total number of times the post was displayed | Measure reach and visibility of content |
| reach | Integer | Number of unique users who saw the post | Assess actual audience size reached |
| likes | Integer | Number of likes/reactions on the post | Basic engagement metric |
| comments | Integer | Number of comments on the post | Measure active user interaction |
| shares | Integer | Number of shares/reposts of the content | Track content virality and amplification |
| saves | Integer | Number of users who saved the post (Instagram/Facebook only) | Measure content value and future reference intent |
| clicks | Integer | Number of clicks on post links or call-to-actions | Track user action and conversion potential |
| video_views | Integer | Number of video views (for video content only) | Measure video content consumption |
| post_cost_cad | Integer | Individual cost for each post in CAD | Calculate cost-effectiveness per post |
| engagement_rate | Float | Percentage of reach that engaged with content | Primary performance indicator |
| cost_per_engagement | Float | Cost divided by total engagements | Measure cost-effectiveness of campaigns |
| click_through_rate | Float | Percentage of reach that clicked on links | Measure call-to-action effectiveness |
| cost_per_click | Float | Cost divided by total clicks | Track cost-efficiency of driving traffic |
| total_engagement | Integer | Sum of likes, comments, and shares | Aggregate engagement measurement |
| engagement_per_dollar | Float | Total engagement divided by post cost | Overall campaign efficiency metric |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("canadian_heritage_events_dataset.csv")
df.head()
```

### Missing Value Analysis

```python
print("\nMissing values check:")
print(df.isnull().sum().sum(), "total missing values")
```

### Data Type Optimization

```python
# Data Type Optimization

print("DATA TYPE CONVERSIONS")


# Convert date columns to datetime for proper time-based analysis
df['event_date'] = pd.to_datetime(df['event_date'])
df['post_date'] = pd.to_datetime(df['post_date'])

print("Converted date columns:")
print("- event_date:", df['event_date'].dtype)
print("- post_date:", df['post_date'].dtype)

# Check if any numeric columns are stored as objects (shouldn't be, but let's verify)
numeric_columns = ['impressions', 'reach', 'likes', 'comments', 'shares', 
                   'saves', 'clicks', 'video_views', 'engagement_rate', 
                   'cost_per_engagement', 'total_engagement']

print("\nNumeric column data types:")
for col in numeric_columns:
    print(f"- {col}: {df[col].dtype}")
    
# If any are objects, convert them
for col in numeric_columns:
    if df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col], errors='coerce')
        print(f"Converted {col} from object to numeric")
```

### Feature Engineering 

```python
print("FEATURE ENGINEERING")

# Add useful derived features for better analysis
df['campaign_phase'] = np.where(df['post_date'] <= df['event_date'], 'Pre-Event', 'Post-Event')
df['weekend_post'] = df['post_date'].dt.dayofweek >= 5  # Saturday=5, Sunday=6
df['month'] = df['post_date'].dt.month_name()

# Calculate engagement distribution
df['likes_ratio'] = df['likes'] / df['total_engagement']
df['comments_ratio'] = df['comments'] / df['total_engagement'] 
df['shares_ratio'] = df['shares'] / df['total_engagement']

print("Added derived features:")
print("- campaign_phase: Pre-Event vs Post-Event timing")
print("- weekend_post: Whether posted on weekend")
print("- month: Month of posting")
print("- engagement ratios: Breakdown of engagement types")

print(f"\nCampaign phase distribution:")
print(df['campaign_phase'].value_counts())

print(f"\nWeekend vs weekday posts:")
print(df['weekend_post'].value_counts())
```
