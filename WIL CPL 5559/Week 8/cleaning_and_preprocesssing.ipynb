```python
import pandas as pd
import numpy as np

# Load raw summary CSV
df = pd.read_csv('cleaned_synthetic_wellness_data.csv')
```

```python
df.head()
```

```python
df.info()
```

```python

# 1. Impute missing values
df['avg_steps_per_event'] = df['avg_steps_per_event'].fillna(0)
mean_ws = df['wellness_score_avg'].mean()
df['wellness_score_avg'] = df['wellness_score_avg'].fillna(mean_ws)
```

```python
# 2. Round error rates to two decimals
df['error_rate_before_%'] = df['error_rate_before_%'].round(2)
df['error_rate_after_%'] = df['error_rate_after_%'].round(2)
```

```python

# 3. Clip numeric ranges for consistency
df['participation_rate_%'] = df['participation_rate_%'].clip(0, 100)
df['unique_event_types'] = df['unique_event_types'].clip(0, 4)
df['mental_health_days_count'] = df['mental_health_days_count'].clip(0, 6)
df['nutrition_sessions_count'] = df['nutrition_sessions_count'].clip(0, 6)
df['milestones_met_count'] = df['milestones_met_count'].clip(0, 10)
df['referred_others'] = df['referred_others'].clip(0, 3)
df['age'] = df['age'].clip(18, 65).astype(int)
```

```python
# 4. Address outliers in productivity delta
mean_delta = df['productivity_delta_tasks'].mean()
std_delta = df['productivity_delta_tasks'].std()
lower, upper = mean_delta - 3*std_delta, mean_delta + 3*std_delta
df['productivity_delta_tasks'] = df['productivity_delta_tasks'].clip(lower, upper)
```

```python

# 5. Standardize categorical fields
df['parent_engagement'] = df['parent_engagement'].str.title()
df['department'] = df['department'].str.capitalize()
```

```python

# 6. Fill lag days and ensure integer type
df['wellness_to_productivity_lag_days'] = df['wellness_to_productivity_lag_days'].fillna(0).astype(int)
```

```python

# 7. Export cleaned dataset to CSV
df.to_csv('cleaned_synthetic_wellness_data.csv', index=False)
```
