```python
import numpy as np
import pandas as pd
```

```python
# 1. Set random seed for reproducibility
np.random.seed(42)
```

```python

# 2. Define parameters
num_employees = 500
start_date = pd.to_datetime('2024-01-01')
end_date = pd.to_datetime('2024-12-31')
date_range = pd.date_range(start_date, end_date, freq='MS')  # Monthly start dates
```

```python

# 3. Generate employee demographic details
employee_ids = [f"E{1000 + i}" for i in range(num_employees)]
ages = np.random.normal(loc=35, scale=10, size=num_employees).astype(int)
departments = np.random.choice(
    ['HR', 'Finance', 'IT', 'Operations'],
    size=num_employees,
    p=[0.2, 0.2, 0.3, 0.3]
)
```

```python
# 4. Designate a control group (20% of employees)
is_control = np.random.choice([True, False], size=num_employees, p=[0.2, 0.8])
```

```python
# 5. Prepare list to collect per-employee summary data
summary_records = []
```

```python
for emp_id, age, dept, ctrl in zip(employee_ids, ages, departments, is_control):
    # Baseline productivity metrics
    tasks_before = int(np.random.normal(20, 5))
    error_before = round(np.random.beta(2, 10) * 10, 2)

    if ctrl:
        # Control group: no wellness events
        summary_records.append({
            'employee_id': emp_id,
            'age': age,
            'department': dept,
            'is_control_group': True,
            'events_attended_count': 0,
            'participation_rate_%': 0.0,
            'unique_event_types': 0,
            'total_steps_logged': 0,
            'avg_steps_per_event': 0.0,
            'wellness_score_avg': np.nan,
            'mental_health_days_count': 0,
            'nutrition_sessions_count': 0,
            'tasks_completed_before': tasks_before,
            'error_rate_before_%': error_before,
            'tasks_completed_after': tasks_before,
            'error_rate_after_%': error_before,
            'milestones_met_count': 0,
            'productivity_delta_tasks': 0,
            'productivity_delta_error_%': 0.0,
            'wellness_to_productivity_lag_days': np.nan,
            'referred_others': np.random.randint(0, 4),
            'parent_engagement': np.random.choice(['Low', 'Medium', 'High'])
        })
        continue

    # Treatment group: simulate wellness events and summary metrics
    num_events = np.random.randint(3, 7)
    event_types = np.random.choice(
        ['Steps Challenge', 'Mental Health Day', 'Nutrition Tracking', 'Yoga Session'],
        size=num_events, p=[0.4, 0.2, 0.3, 0.1]
    )
    steps_list = [int(np.random.normal(8000, 2000)) if etype == 'Steps Challenge' else np.nan for etype in event_types]
    wellness_list = [round(np.random.uniform(1, 5), 2) for _ in event_types]
    mental_days = sum(1 for etype in event_types if etype == 'Mental Health Day')
    nutrition_count = sum(1 for etype in event_types if etype == 'Nutrition Tracking')

    # Post-participation productivity metrics
    improvement = int(np.random.normal(2, 3))
    tasks_after = max(0, tasks_before + improvement)
    error_after = round(max(0, error_before - np.random.uniform(0, 2)), 2)

    # Summary-level calculations
    events_attended = num_events
    participation_rate = round((events_attended / 6) * 100, 2)
    unique_types = len(set(event_types))
    total_steps = int(np.nansum(steps_list))
    avg_steps = round(np.nanmean(steps_list), 2)
    avg_wellness = round(np.mean(wellness_list), 2)
    lag_days = int((pd.to_datetime('2024-12-31') - pd.to_datetime('2024-01-01')).days * (events_attended/12))

    summary_records.append({
        'employee_id': emp_id,
        'age': age,
        'department': dept,
        'is_control_group': False,
        'events_attended_count': events_attended,
        'participation_rate_%': participation_rate,
        'unique_event_types': unique_types,
        'total_steps_logged': total_steps,
        'avg_steps_per_event': avg_steps,
        'wellness_score_avg': avg_wellness,
        'mental_health_days_count': mental_days,
        'nutrition_sessions_count': nutrition_count,
        'tasks_completed_before': tasks_before,
        'error_rate_before_%': error_before,
        'tasks_completed_after': tasks_after,
        'error_rate_after_%': error_after,
        'milestones_met_count': np.random.randint(0, 4),
        'productivity_delta_tasks': tasks_after - tasks_before,
        'productivity_delta_error_%': round(error_before - error_after, 2),
        'wellness_to_productivity_lag_days': lag_days,
        'referred_others': np.random.randint(0, 4),
        'parent_engagement': np.random.choice(['Low', 'Medium', 'High'])
    })
```

```python
# 6. Create summary DataFrame
summary_df = pd.DataFrame(summary_records)
```

```python
# 7. Export only the summary dataset to CSV
summary_df.to_csv('synthetic_summary.csv', index=False)
```
